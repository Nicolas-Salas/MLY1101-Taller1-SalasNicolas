````python
# Librerías base
import requests
import json
import pandas as pd
from pathlib import Path
import time

OUTPUT_DIR = Path('datasets')
OUTPUT_DIR.mkdir(exist_ok=True)

print(f'Carpeta de salida: {OUTPUT_DIR.resolve()}')
````

# Fuente 1 — API 1

**Nombre de la API:**  
**Documentación:**  
**Endpoint utilizado:**  
**Descripción de los datos:**  
**Relación con el objetivo:**  


````python
# CONFIGURACIÓN API 1
API1_URL = 'https://api.coinlore.net/api/tickers/?start=0&limit=200'
API1_MIN_REGISTROS = 200

api1_headers = {}
api1_params = {}
````

````python
# CONSUMO API 1

api1_registros = []

response = requests.get(API1_URL, headers=api1_headers, params=api1_params, timeout=30)
print('Status API 1:', response.status_code)
response.raise_for_status()

data = response.json()

api1_registros = data['data']

print('Registros API 1:', len(api1_registros))
````

### Paginación API 1

Si la API utiliza paginación, reemplace el bloque anterior por un ciclo que continúe hasta obtener al menos 200 registros o hasta que no existan más páginas.


````python
api1_registros = []
start_index = 0
limit = 100

while len(api1_registros) < API1_MIN_REGISTROS:
    params = {
        'start': start_index,
        'limit': limit
    }

    response = requests.get(API1_URL, params=params, timeout=30)
    response.raise_for_status()
    data = response.json()
    nuevos = data['data']

    if not nuevos:
        break

    api1_registros.extend(nuevos)
    start_index += limit

print('Status API 1:', response.status_code)
print('Registros API 1:', len(api1_registros))
````

````python
# GUARDAR DATASET API 1
API1_ARCHIVO = OUTPUT_DIR / 'dataset_api_1.json'

with open(API1_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api1_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API1_ARCHIVO)
print('Registros guardados:', len(api1_registros))
````

# Fuente 2 — API 2

**Nombre de la API:**  
**Documentación:**  
**Endpoint utilizado:**  
**Descripción de los datos:**  
**Relación con el objetivo:**  


````python
# CONFIGURACIÓN API 2
API2_URL = 'https://api.github.com/repositories'
API2_MIN_REGISTROS = 200

api2_headers = {"User-Agent": "Mozilla/5.0"}
api2_params = {}
````

````python
# CONSUMO API 2
api2_registros = []
since_id = 0

while len(api2_registros) < API2_MIN_REGISTROS:
    api2_params = {'since': since_id}
    response = requests.get(API2_URL, headers=api2_headers, params=api2_params, timeout=30)
    response.raise_for_status()
    nuevos_registros = response.json()

    if not nuevos_registros:
        break

    api2_registros.extend(nuevos_registros)
    since_id = api2_registros[-1]['id']
    time.sleep(1)

print('Status API 2:', response.status_code)
print('Registros API 2:', len(api2_registros))
````

````python
# GUARDAR DATASET API 2
API2_ARCHIVO = OUTPUT_DIR / 'dataset_api_2.json'

with open(API2_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api2_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API2_ARCHIVO)
print('Registros guardados:', len(api2_registros))
````

````python
# CONFIGURACIÓN API 3
API3_URL = 'https://services.nvd.nist.gov/rest/json/cves/2.0'
API3_MIN_REGISTROS = 200

api3_headers = {"User-Agent": "Mozilla/5.0"}
api3_params = {"resultsPerPage": 200}
````
